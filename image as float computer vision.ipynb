```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage import data,img_as_float
from scipy.ndimage import median_filter,uniform_filter,gaussian_filter
```

```python
img=data.chelsea()
img.shape
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
img=data.astronaut()
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.subplot(1,2,1)
plt.imshow(img)
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(gray,cmap='gray')
plt.axis('off')
```

```python
from skimage.util import random_noise
img=data.camera()
img=img_as_float(img)
gue_moise=random_noise(img,mode='gaussian')
salt_moise=random_noise(img,mode='s&p') 
spe_moise=random_noise(img,mode='speckle') 

g8=(gue_moise*255).astype(np.uint8)
sp8=(salt_moise*255).astype(np.uint8)
spe8=(spe_moise*255).astype(np.uint8)

mean_f=uniform_filter(sp8,size=5)
median_f=median_filter(sp8,size=5)
gaussian_f=gaussian_filter(sp8,sigma=2)


plt.subplot(2,3,1)
plt.imshow(g8,cmap='gray')
plt.axis('off')
plt.subplot(2,3,2)
plt.imshow(sp8,cmap='gray')
plt.axis('off')
plt.subplot(2,3,3)
plt.imshow(spe8,cmap='gray')
plt.axis('off')
plt.subplot(2,3,4)
plt.imshow(mean_f,cmap='gray')
plt.axis('off')
plt.subplot(2,3,5)
plt.imshow(median_f,cmap='gray')
plt.axis('off')
plt.subplot(2,3,6)
plt.imshow(gaussian_f,cmap='gray')
plt.axis('off')
```
